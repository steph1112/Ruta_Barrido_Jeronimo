```python
import gpxpy
import folium

# Función para leer rutas GPX
def leer_ruta_gpx(archivo_gpx):
    with open(archivo_gpx, 'r') as gpx_file:
        gpx = gpxpy.parse(gpx_file)
        coordenadas = []
        for track in gpx.tracks:
            for segment in track.segments:
                for point in segment.points:
                    coordenadas.append((point.latitude, point.longitude))
        return coordenadas

# Cargar las rutas GPX
ruta_alejandrina = leer_ruta_gpx("C:/Users/USER/Workspace/Ruta_Barrido_Jeronimo/Ruta_Barrido_Alejandrina.gpx")
ruta_irene = leer_ruta_gpx("C:/Users/USER/Workspace/Ruta_Barrido_Jeronimo/Ruta_Barrido_Irene.gpx")

# Crear el mapa centrado en San Jerónimo
mapa = folium.Map(location=[-13.546, -71.894], zoom_start=16)

# Agregar las rutas al mapa con diferentes colores
folium.PolyLine(ruta_alejandrina, color="blue", weight=5, opacity=0.7, tooltip="Ruta Alejandrina").add_to(mapa)
folium.PolyLine(ruta_irene, color="red", weight=5, opacity=0.7, tooltip="Ruta Irene").add_to(mapa)

# Puntos de la Ruta de Mojompata 1
puntos_mojompata1 = [
    ([-13.537732, -71.905653], "Punto de Inicio - Mojompata 1", "green", "play"),
    ([-13.546373, -71.896189], "Punto Final - Mojompata 1", "purple", "stop"),
]

for coord, nombre, color, icon in puntos_mojompata1:
    folium.Marker(location=coord, popup=nombre, icon=folium.Icon(color=color, icon=icon)).add_to(mapa)

# Puntos críticos de Mojompata 1
puntos_criticos_mojompata1 = [
    (-13.539294, -71.905011, "Punto Crítico 1 - Mojompata 1"),
    (-13.539735, -71.904186, "Punto Crítico 2 - Mojompata 1"),
    (-13.542143, -71.901977, "Punto Crítico 3 - Mojompata 1"),
    (-13.544976, -71.897424, "Punto Crítico 4 - Mojompata 1"),
    (-13.545647, -71.896850, "Punto Crítico 5 - Mojompata 1"),
]

for lat, lon, nombre in puntos_criticos_mojompata1:
    folium.Marker(location=[lat, lon], popup=nombre, icon=folium.Icon(color="red", icon="exclamation-triangle")).add_to(mapa)

# Desmontes de Mojompata 1
desmontes_mojompata1 = [
    (-13.540313, -71.904837, "Desmonte 1 - Mojompata 1"),
    (-13.539735, -71.904186, "Desmonte 2 - Mojompata 1"),
]

for lat, lon, nombre in desmontes_mojompata1:
    folium.Marker(location=[lat, lon], popup=nombre, icon=folium.Icon(color="orange", icon="trash")).add_to(mapa)

# Puntos de la Ruta de Mojompata 2
puntos_mojompata2 = [
    ([-13.5453310, -71.8941340], "Punto de Inicio - Mojompata 2", "green", "play"),
    ([-13.5468690, -71.8953850], "Punto Final - Mojompata 2", "purple", "stop"),
]

for coord, nombre, color, icon in puntos_mojompata2:
    folium.Marker(location=coord, popup=nombre, icon=folium.Icon(color=color, icon=icon)).add_to(mapa)

# Puntos críticos de Mojompata 2
puntos_criticos_mojompata2 = [
    (-13.547632, -71.892782, "Punto Crítico 1 - Mojompata 2"),
    (-13.547173, -71.89478, "Punto Crítico 2 - Mojompata 2"),
    (-13.548021, -71.890831, "Punto Crítico 3 - Mojompata 2"),
    (-13.547473, -71.89266, "Punto Crítico 4 - Mojompata 2"),
]

for lat, lon, nombre in puntos_criticos_mojompata2:
    folium.Marker(location=[lat, lon], popup=nombre, icon=folium.Icon(color="red", icon="exclamation-triangle")).add_to(mapa)

# Agregar leyenda flotante diferenciando puntos críticos y desmontes con nombres de los monitores
legend_html = """
<div style="position: fixed; 
            bottom: 50px; left: 50px; width: 320px; height: 260px; 
            background-color: white; z-index:9999; padding: 10px;
            border-radius: 10px; box-shadow: 2px 2px 5px grey; font-size:14px;">
    <b>Ruta de Barrido - Mojompata 1 & 2</b><br>
    <b>Monitores: Alejandro Bernable & Nicole Román</b><br><br>
    <span style="color:green;">&#9654;</span> Inicio de Ruta<br>
    <span style="color:purple;">&#9632;</span> Final de Ruta<br>
    <span style="color:red;">&#9888;</span> Puntos Críticos<br>
    <span style="color:orange;">&#128465;</span> Desmontes<br>
    <span style="color:blue;">&#128200;</span> Ruta Alejandrina<br>
    <span style="color:red;">&#128201;</span> Ruta Irene<br>
</div>
"""
    
    
mapa.get_root().html.add_child(folium.Element(legend_html))

# Guardar el mapa en un archivo HTML
mapa.save("Ruta Mojompata.html")

# Mostrar el mapa en Jupyter Notebook (si aplica)
mapa
```
